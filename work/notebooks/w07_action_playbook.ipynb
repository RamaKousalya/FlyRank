## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Find the CSV
files = list(Path(".").rglob("*.csv"))
print(files)

# Load Week-5 queue if available
queue_files = [p for p in files if "baseline_action_score" in p.name]

if queue_files:
    queue = pd.read_csv(queue_files[0])
else:
    # Load original dataset
    df = pd.read_csv(files[0], header=None)

    # First column = target, remaining columns = features
    X = df.iloc[:, 1:]
    y = df.iloc[:, 0]

    # Simple ranked decision-support queue
    queue = pd.DataFrame({
        "item": range(len(df)),
        "score": 0
    })

    queue["reason_code"] = "REVIEW"
    queue["action_label"] = "MONITOR"

queue = queue.sort_values("score", ascending=False)

display(queue.head(10))
print("Rows:", len(queue))
```
